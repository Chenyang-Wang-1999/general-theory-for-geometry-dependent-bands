```python
import numpy as np
```

```python
def complex_to_sphere(beta):
    return np.array([2*beta.real/(1 + abs(beta)**2), 2*beta.imag/(1+abs(beta)**2), (abs(beta)**2 - 1)/(abs(beta)**2 + 1)])
```

```python
a = 0.9
b = 0.8
sp_a = complex_to_sphere(a)
sp_b = complex_to_sphere(b)
np.linalg.norm(sp_a - sp_b)
```
