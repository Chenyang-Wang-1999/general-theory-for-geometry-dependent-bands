```python
class MyClass:
    a:list
    b:list
    def __init__(self, a, b) -> None:
        self.a = a
        self.b = b
    def __getitem__(self, key:int):
        if(key>=len(self.a)):
            return self.b[key-len(self.a)]
        else:
            return self.a[key]
```

```python
my_class = MyClass([1,2], [3,4,5])
```

```python
my_class[4] = 1
```
