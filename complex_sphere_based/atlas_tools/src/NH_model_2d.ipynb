# Test beta dependent spectrum

```python
Jx1 = 1
Jx2 = 2
Jy1 = 0.9j
Jy2 = 0.9j
```

```python
beta1 = np.exp(1j * np.linspace(0, 2*np.pi, 100))
phi20 = np.pi/3
E1 = np.cos(phi20/2) * np.sqrt((Jx2 * beta1 + Jy1) * (Jx1 / beta1 + Jy2))
E2 = - E1
```

```python
plt.plot(E1.real, E1.imag, '.')
plt.plot(E2.real, E2.imag, '.')
E0 = np.cos(phi20/2) * np.sqrt((Jx2 * beta1[0] + Jy1) * (Jx1 / beta1[0] + Jy2))
plt.plot(E0.real, E0.imag, 'rx')
plt.show()
```
