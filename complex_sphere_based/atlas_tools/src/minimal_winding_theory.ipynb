```python
N = 50
k = pi/3
theta = np.linspace(0, 2 * pi, 1000)
phi_list = np.linspace(1/(N+1), 1 - 1/(N+1), N) * pi
for phi in phi_list:
    r = np.tan(phi/2)
    E = r * np.exp(1j * theta) + (
        np.exp(-1j * theta) + np.exp(1j * theta) * exp(-1j * k) + 
        np.exp(-2j * theta) * exp(1j * k)
    )/r
    plt.clf()
    plt.title("%f"%(r))
    plt.plot(E.real, E.imag)
    plt.pause(1)
    
plt.show()
```

# Y major GBZ

```python
import BerryPy.TightBinding as tb
```

```python
def y_major_1d_model(beta2:complex):
    dim = 1
    intra_cell = []
    inter_cell = [
        [0, 0, 1, (1,)],
        [0, 0, 1 + 1/beta2, (-1,)],
        [0, 0, beta2, (2,)]
    ]
    return tb.TightBindingModel(dim, 1, [[1]], 
            intra_cell, inter_cell)
```

```python
# test y major model
r = 1.1
theta = pi/3
beta2 = r * np.exp(1j * theta)
model = y_major_1d_model(beta2)
GBZ_result = model.get_GBZ()
```

```python
flag, E, beta1 = GBZ_result
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(beta1.real, beta1.imag, E.real, '.')
plt.show()
```

```python
r = 1.1
theta = np.linspace(0, 2 * pi, 50)
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
for curr_theta in theta:
    beta2 = r * np.exp(1j * curr_theta)
    model = y_major_1d_model(beta2)
    flag, E, beta1 = model.get_GBZ()
    ax.plot(beta1.real, beta1.imag, E.real, '.')
    plt.pause(1)
plt.show()
```
