```python
data = sio.loadmat("data/GBZ_char_poly_5.mat")
```

```python
coeffs = data['coeffs'].flatten()
orders = data['orders'].flatten()
```

```python
char_poly = csolver.CLaurant(3)
char_poly.set_Laurant_by_terms(csolver.CScalarVec(coeffs), csolver.CIntVec(orders))
```

```python
char_poly.num.to_str(csolver.CStrVec(['x', 'y','z']))
```

```python
continuous_points = []
continuous_params = []
for j in range(len(bug_points)):
    p_sample = bug_points[j]
    param_sample = bug_params[j]
    winding_solver = csolver.GBZConditionChecker2D(p_sample[:3], param_sample[1], char_poly, 1, 0.1)
    result = winding_solver.eval_param_path(p_sample[1]*np.exp(1j * np.linspace(0,2*np.pi, 100)), p_sample[[0,2]], 
                                                N_points_loop=200, N_points_detour=50, theta_detour=1e-2*np.pi)
    if(not (result is None)):
        continuous_points.append(p_sample)
        continuous_params.append(param_sample)
```
