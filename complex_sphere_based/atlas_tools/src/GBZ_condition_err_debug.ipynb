```python
normal_points_selected = []
normal_params_selected = []
for j, p in enumerate(bug_points) :
    if((p[0].real < -1.4) and (p[0].real > -2) and (p[0].imag > 0.2) and (p[0].imag < 1.0)):
        normal_points_selected.append(p)
        normal_params_selected.append(bug_params[j])
    else:
        print(p, j)
```

```python
bug_points_selected = []
bug_params_selected = []
for j, p in enumerate(bug_points) :
    if((p[0].real < 2.0) and (p[0].real > 0.8) and (p[0].imag > -0.05) and (p[0].imag < 0.15)):
        bug_points_selected.append(p)
        bug_params_selected.append(bug_params[j])
    else:
        print(p, j)
```
