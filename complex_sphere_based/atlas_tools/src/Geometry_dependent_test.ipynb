```python
plt.figure()
ax = plt.gca()
for j in range(50):
    if (os.path.exists("data/GBZ_data_1/y_major_%d.pkl"%(j))):
        with open("data/GBZ_data_1/y_major_%d.pkl"%(j), "rb") as fp:
            all_segments = pickle.load(fp)
        for seg in all_segments:
            ax.plot(seg[:,0].real, seg[:,0].imag, 'b.')
        # plt.pause(0.01)
        # err_flag = int(input("%d"%(j)))
        # if(err_flag):
        #     print("non GBZ point: %d"%(j))
    else:
        print("solution error: %d"%(j))
ax.set_xlabel("$\\mathrm{Re}(E)$")
ax.set_ylabel("$\\mathrm{Im}(E)$")
ax.set_xlim([-10,10])
ax.set_ylim([-3,3])
plt.savefig("Figures/y_major_1.svg")
```
